```python
import pandas as pd

df = pd.read_excel("/content/Dot_and_Key_Brand_Launch_Intelligence_Dataset.xlsx")

print("Dataset loaded!")
```

```python
import sqlite3
conn = sqlite3.connect("/content/dot_key_marketing.db")
df.to_sql("orders", conn, if_exists='replace', index=False)
print('SQL database created successfully')
```

```python
query= """
SELECT Marketing_Channel,
SUM(Revenue) AS Total_Revenue,
    SUM(Ad_Spend) AS Total_Ad_Spend,
    SUM(Profit) AS Total_Profit
FROM orders
GROUP BY Marketing_Channel
ORDER BY Total_Revenue DESC
"""
Channel_revenue = pd.read_sql_query(query,conn)
Channel_revenue
```

```python
from numpy import prod
Query ="""
SELECT Product,Category,
SUM(Revenue) AS Total_Revenue,
SUM(Ad_Spend) AS Total_Ad_Spend,
SUM(Profit) AS Total_Profit
FROM orders
GROUP BY Product,Category
ORDER BY Total_Revenue DESC
"""
product_analysis = pd.read_sql_query(Query,conn)
product_analysis
```

```python
query ="""
SELECT Customer_segment,
COUNT(DISTINCT Customer_ID) AS Unique_Customers,
SUM(Revenue) AS Total_Revenue,
    SUM(Profit) AS Total_Profit,
    AVG(Revenue) AS Average_Order_Value
FROM orders
GROUP BY Customer_segment
ORDER BY Unique_Customers DESC
"""
Segment_analysis = pd.read_sql_query(query,conn)
Segment_analysis
```

```python
query = """
SELECT
    Marketing_Channel,
    SUM(Revenue) AS Total_Revenue,
    SUM(Ad_Spend) AS Total_Ad_Spend,
    ROUND(SUM(Revenue)/SUM(Ad_Spend),2) AS ROAS
FROM orders
GROUP BY Marketing_Channel
ORDER BY ROAS DESC;
"""

roas_analysis = pd.read_sql_query(query, conn)

roas_analysis
```

```python
query = """
SELECT
    Campaign,
    SUM(Impressions) AS Total_Impressions,
    SUM(Clicks) AS Total_Clicks,
    SUM(Leads) AS Total_Leads,
    SUM(Conversions) AS Total_Conversions,
   ROUND(SUM(Conversions)*100.0/SUM(Clicks),2) AS Conversion_Rate
   FROM orders
   GROUP BY Campaign
   ORDER BY Conversion_Rate DESC
"""

campaign_analysis = pd.read_sql_query(query, conn)

campaign_analysis
```

```python
product_profit=df.groupby('Product')['Profit'].sum().sort_values(ascending=False)
product_profit
```

```python
segment_profit = (
    df.groupby("Customer_Segment")["Profit"]
    .sum()
    .sort_values(ascending=False)
)

segment_profit
```

```python
channel_performance=df.groupby('Marketing_Channel')[['Revenue','Ad_Spend','Profit']].sum()
channel_performance['ROAS']=round(channel_performance['Profit']/channel_performance['Ad_Spend'],2)
channel_performance.sort_values(by='ROAS',ascending=False)
```

```python
city_performance = (
    df.groupby("City")[["Revenue", "Profit"]]
    .sum()
    .sort_values("Revenue", ascending=False)
)

city_performance.head(10)
```

```python
customer_type_performance = (
    df.groupby("Customer_Type")[["Revenue", "Profit", "Quantity"]]
    .sum()
    .sort_values("Revenue", ascending=False)
)

customer_type_performance
```

```python
campaign_funnel = (
    df.groupby("Campaign")[["Impressions", "Clicks", "Leads", "Conversions"]]
    .sum()
    .sort_values("Conversions", ascending=False)
)

campaign_funnel
```

```python
campaign_funnel["CTR"] = (
    campaign_funnel["Clicks"] / campaign_funnel["Impressions"] * 100
).round(2)

campaign_funnel["Lead_Rate"] = (
    campaign_funnel["Leads"] / campaign_funnel["Clicks"] * 100
).round(2)

campaign_funnel["Conversion_Rate"] = (
    campaign_funnel["Conversions"] / campaign_funnel["Clicks"] * 100
).round(2)

campaign_funnel
```

```python
loss_by_product = (
    df[df["Profit"] < 0]
    .groupby("Product")["Profit"]
    .sum()
    .sort_values()
)

loss_by_product
```
